```python
import pandas as pd
import altair as alt

df = pd.read_csv("calendar_clean.csv")


df25 = df[df["year"] == 2025]


medians = df25.groupby(["month", "park_name"])["crowd_level_pct"].median().reset_index()

medians["quietest"] = medians["crowd_level_pct"] == medians.groupby("month")["crowd_level_pct"].transform("min")

months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
medians["month_name"] = medians["month"].map({i + 1: m for i, m in enumerate(months)})

chart = alt.Chart(medians).mark_circle(size=200).encode(
    x=alt.X("month_name", sort=months, title="Month"),
    y=alt.Y("crowd_level_pct", title="Median crowd score"),
    color="park_name"
).properties(
    title="Which Park Is Quietest Each Month? (2025)",
    width=600,
    height=400
)

chart
```
