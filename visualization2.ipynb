```python
import pandas as pd
import altair as alt

rides = pd.read_csv("rides_clean_2025.csv")

category_select = alt.selection_point(fields=["category"], bind="legend")

chart = alt.Chart(rides).mark_circle(size=200).encode(
    x=alt.X("avg_wait:Q", title="Average wait (minutes)"),
    y=alt.Y("max_wait:Q", title="Maximum wait (minutes)"),
    color=alt.Color("category:N", title="Category"),
    opacity=alt.condition(category_select, alt.value(0.9), alt.value(0.1)),
    tooltip=[
        alt.Tooltip("ride_name:N", title="Ride"),
        alt.Tooltip("park_name:N", title="Park"),
        alt.Tooltip("avg_wait:Q", title="Average wait (min)"),
        alt.Tooltip("max_wait:Q", title="Maximum wait (min)"),
    ],
).add_params(
    category_select
).properties(
    title="Which Rides Are Worth the Wait? (2025)",
    width=600,
    height=400
)

chart
```
